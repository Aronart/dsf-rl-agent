# Stock Data Download and Preparation (FMP API)

This notebook implements the data download and preparation process for stock analysis using the **Financial Modeling Prep (FMP) API**.

### 1. Setup
- **API Key**: Make sure you have a `.env` file in your project root with your FMP API key: `FMP_API_KEY=your_key_here`.
- **Dependencies**: This script requires `pandas`, `yfinance`, `requests`, and `python-dotenv`.

### 2. Process
The script performs the following steps:
1. **Downloads historical stock prices** from Yahoo Finance.
2. **Fetches company news** from the FMP API, filtering for articles published outside of market hours.
3. **Retrieves company press releases** from the FMP API.
4. **Combines and formats** the data into a single CSV file.

### 3. Output
The final output is a CSV file named `{symbol}_{start_date}_{end_date}.csv` with the following columns:
- `Date`
- `Adj Close Price`, `Returns`, `Bin Label`
- `News` (a JSON string of relevant news articles)
- `PressReleases` (a JSON string of relevant press releases)

```python
import os
import pandas as pd
import json
from prepare_data_fmp import prepare_data_for_symbol, get_company_profile
```

```python
symbols = ["AAPL", "NFLX", "MSFT", "CRM", "AMZN"]
prepared = {}
start_date = "2015-01-01"
end_date = "2025-05-31"
for stock_symbol in symbols:
    data_dir = f"data/{stock_symbol}_{start_date}_{end_date}"
    os.makedirs(data_dir, exist_ok=True)
    print(f"Output directory created at: {data_dir}")

    prepared[stock_symbol] = prepare_data_for_symbol(stock_symbol, data_dir, start_date, end_date)
```
